```python
import pandas as pd

print("Pandas Imported Successfully")
```

```python
df = pd.read_csv("StudentsPerformance.csv")

print("Dataset Loaded Successfully")
```

```python
print(df.head())
```

```python
math_scores = df["math score"].head(10).tolist()

print("First 10 Math Scores:")
print(math_scores)
```

```python
def grade(m):
    if m >= 80:
        return "A"
    elif m >= 60:
        return "B"
    elif m >= 40:
        return "C"
    else:
        return "F"
```

```python
print("Grade of 90:", grade(90))
print("Grade of 65:", grade(65))
print("Grade of 45:", grade(45))
print("Grade of 30:", grade(30))
```

```python
print("Student Marks and Grades:")

for score in math_scores:
    print("Math Score:", score, "Grade:", grade(score))
```

```python
passed = 0

for score in math_scores:
    if score >= 40:
        passed += 1

print("Total Passed Students:", passed)
```

```python
print("Highest Score:", max(math_scores))
print("Lowest Score:", min(math_scores))
```
