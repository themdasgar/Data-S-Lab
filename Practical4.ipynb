```python
import pandas as pd
from scipy import stats

df = pd.read_csv("StudentsPerformance.csv")
print(df.head())
```

```python
completed = df[df["test preparation course"] == "completed"]["math score"]
none = df[df["test preparation course"] == "none"]["math score"]

t, p1 = stats.ttest_ind(completed, none)

print("T-Test:", t)
print("P-Value:", p1)
```

```python
table = pd.crosstab(df["gender"], df["test preparation course"])

chi, p2, dof, expected = stats.chi2_contingency(table)

print("Chi-Square:", chi)
print("P-Value:", p2)
```

```python
groups = [g["math score"].values for name, g in df.groupby("parental level of education")]

f, p3 = stats.f_oneway(*groups)

print("ANOVA:", f)
print("P-Value:", p3)
```

```python
for name, p in [("T-Test", p1), ("Chi-Square", p2), ("ANOVA", p3)]:
    print(name, ":", "Reject H0" if p < 0.05 else "Fail to Reject H0")
```
