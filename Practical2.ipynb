```python
import pandas as pd
import numpy as np

print("Libraries Imported Successfully")
```

```python
df = pd.read_csv("StudentsPerformance.csv")

print("Dataset Loaded Successfully")
```

```python
print("Dataset Shape:", df.shape)
```

```python
print(df.head())
```

```python
df.info()
```

```python
df["total"] = (
    df["math score"] +
    df["reading score"] +
    df["writing score"]
)

print(df[["gender", "math score", "reading score", "writing score", "total"]].head())
```

```python
average_total = df.groupby("gender")["total"].mean()

print("Average Total Marks by Gender:")
print(average_total)
```

```python
highest_student = df.loc[df["total"].idxmax()]

print("Student with Highest Total:")
print(highest_student)
```
