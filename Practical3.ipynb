```python

import pandas as pd
import numpy as np
from scipy import stats
```

```python

df = pd.read_csv("Iris.csv")
```

```python
mean_value = df["PetalLengthCm"].mean()

print("Mean:", mean_value)
```

```python
median_value = df["PetalLengthCm"].median()

print("Median:", median_value)
```

```python
mode_value = df["PetalLengthCm"].mode()

print("Mode:")
print(mode_value)
```

```python
std_value = df["PetalLengthCm"].std()

print("Standard Deviation:", std_value)
```

```python
species_mean = df.groupby("Species")["PetalLengthCm"].mean()

print("Average Petal Length by Species:")
print(species_mean)
```

```python


print("Mean Petal Length of Each Species:")

print(
    df.groupby("Species")["PetalLengthCm"].mean()
)
```
